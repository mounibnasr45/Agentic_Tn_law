```python
# test_direct_llm_standalone.py
import os
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage

# --- Configuration Directe (NE PAS UTILISER EN PRODUCTION AVEC DES CLÉS EN DUR) ---
# Remplacez par VOTRE clé API OpenRouter
OPENROUTER_API_KEY = "REDACTED-LEAKED-KEY-REVOKED"
OPENROUTER_API_BASE = "https://openrouter.ai/api/v1"
MODEL_NAME = "deepseek/deepseek-chat" # Le modèle que curl a réussi à interroger

# En-têtes recommandés par OpenRouter
# Votre nom d'application ou de projet
APP_TITLE = "TunLawAgentStandaloneTest"
# L'URL où votre application est hébergée (même si c'est localhost pour le développement)
HTTP_REFERER = "http://localhost/tun_law_agent_test" # Vous pouvez ajuster ceci

def main():
    print("--- Test direct de ChatOpenAI avec OpenRouter (configuration en dur) ---")

    if not OPENROUTER_API_KEY or not OPENROUTER_API_KEY.startswith("sk-or-v1-"):
        print("ERREUR CRITIQUE: OPENROUTER_API_KEY n'est pas définie correctement dans ce script.")
        print("Veuillez la remplacer par votre clé OpenRouter valide.")
        return

    print(f"Utilisation du modèle : {MODEL_NAME}")
    print(f"URL de base OpenRouter : {OPENROUTER_API_BASE}")
    print(f"Clé API (tronquée) : ...{OPENROUTER_API_KEY[-6:]}")
    print(f"En-tête X-Title : {APP_TITLE}")
    print(f"En-tête HTTP-Referer : {HTTP_REFERER}")

    try:
        # Initialisation du client ChatOpenAI
        llm = ChatOpenAI(
            model_name=MODEL_NAME,
            openai_api_key=OPENROUTER_API_KEY, # Doit être la clé brute "sk-or-v1-..."
            openai_api_base=OPENROUTER_API_BASE,
            temperature=0.1,
            max_tokens=256,
            default_headers={ # Ajout des en-têtes recommandés
                "X-Title": APP_TITLE,
                "HTTP-Referer": HTTP_REFERER,
                # "Authorization": f"Bearer {OPENROUTER_API_KEY}" # Normalement PAS nécessaire ici,
                                                                # la bibliothèque openai s'en charge
                                                                # en se basant sur openai_api_key.
            }
            # Vous pouvez ajouter ici request_timeout=60 si besoin
        )

        print("\nInstance ChatOpenAI créée. Envoi de la requête...")

        # Préparation des messages pour la conversation
        messages = [
            SystemMessage(content="Vous êtes un assistant IA concis et direct."),
            HumanMessage(content="Bonjour ! Pouvez-vous me dire quelle heure il est conceptuellement ?")
        ]

        # Appel au LLM
        response = llm.invoke(messages)

        print("\n--- Réponse du LLM ---")
        if hasattr(response, 'content'):
            print(response.content)
        else:
            print(response) # Au cas où la structure de réponse serait inattendue
        print("----------------------")
        print("\nTest réussi si vous voyez une réponse cohérente ci-dessus !")

    except Exception as e:
        print(f"\n--- ERREUR lors de l'appel au LLM ---")
        print(f"Une erreur est survenue : {type(e).__name__} - {e}")
        # Tenter d'afficher plus de détails si c'est une erreur HTTP de la bibliothèque openai
        if hasattr(e, 'response') and hasattr(e.response, 'json'):
            try:
                error_detail = e.response.json()
                print(f"Détails de l'erreur API : {error_detail}")
            except:
                pass # Si .json() échoue aussi
        elif hasattr(e, 'status_code'): # Pour les erreurs httpx directes si openai < 1.0
             print(f"Status Code: {e.status_code}")


if __name__ == "__main__":
    main()
```
